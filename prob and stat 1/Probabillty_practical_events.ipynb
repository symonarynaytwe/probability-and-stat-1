```python
sample_space = {1,2,3,4,5,6}
E = {2,4,6}
A = {x for x in sample_space if x>3}
b = {x for x in sample_space if x<7}

print(f"Event E : {E}")
```

we use the fractions package to avoid decimals

```python
from fractions import Fraction
```

function used for finding probability

```python
def prob(event, sample_space):
    event = set(event) & set(sample_space)
    return Fraction(len(event), len(sample_space))
```

```python
## the probability of getting an even number.
print(f"the probability of getting an even number: {prob(E,sample_space)}")
```

```python
## the probability of getting an even number.
print(f"the probability of getting an even number: {prob(b,sample_space)}")
```

```python
## the probability of getting an even number.
print(f"the probability of getting an even number: {prob(A,sample_space)}")
```
