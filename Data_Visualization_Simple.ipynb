Same 8 charts as the original notebook, using shorter pandas built-in plotting where possible (heatmap and the interactive plot still need seaborn/plotly).

```python
# Import libraries
import pandas as pd
import seaborn as sns
import plotly.express as px
```

```python
# our dataset from kaggle
df = pd.read_csv("StudentsPerformance.csv")
df.head()
```

## 1. Bar Chart

```python
df.groupby("parental level of education")["math score"] \
    .mean().sort_values() \
    .plot.barh(title="Average Math Score by Parental Education Level")
```

## 2. Line Chart

```python
df["math score"].sort_values().reset_index(drop=True) \
    .plot(title="Sorted Math Scores (Trend Line)")
```

## 3. Histogram

```python
df["writing score"].plot.hist(bins=20, title="Distribution of Writing Scores")
```

## 4. Scatter Plot

```python
df.plot.scatter(x="math score", y="reading score", title="Math Score vs Reading Score")
```

## 5. Box Plot

```python
df.boxplot(column="math score", by="test preparation course")
```

## 6. Pie Chart

```python
df["race/ethnicity"].value_counts().plot.pie(
    autopct="%1.1f%%", title="Proportion of Students by Group"
)
```

## 7. Heatmap

```python
sns.heatmap(
    df[["math score", "reading score", "writing score"]].corr(),
    annot=True, cmap="Blues", vmin=0, vmax=1
)
```

## 8. Interactive Plot with Plotly

```python
fig = px.scatter(
    df, x="math score", y="writing score", color="gender",
    title="Math Score vs Writing Score (Interactive)"
)
fig.show()
```
